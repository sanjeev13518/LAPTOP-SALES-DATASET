## One Sample T-test

```python
# One Sample t-test

from scipy import stats

sample_data=Laptop_df1['Price']

population_mean=70000

# One sample t-test
t_statistic, p_value = stats.ttest_1samp(sample_data, population_mean)

# Output the results
print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")

# Determine the result
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: The sample mean is not significantly different from the population mean.")
```

#- A one-sample t-test was conducted to determine whether the mean laptop price differs from the **population mean of ₹70,000.**

#- So Reject the null hypothesis: The sample mean is significantly different from the population mean.

## Chi Square Test

```python
# Chi Square Test

import seaborn as sns
from scipy.stats import chi2_contingency

#Company vs TypeName
contingency_table = pd.crosstab(Laptop_df1['Company'], Laptop_df1['TypeName'])
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("contingency_table", contingency_table)
print("Chi-square statistic:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)

if p < 0.05:
    print("Reject H0 → There is a significant relationship")
else:
    print("Fail to Reject H0 → No significant relationship")
```

**We applied a Chi-Square test of independence between Laptop Company and Laptop TypeName.**

- Null Hypothesis (H0): Laptop brand (Company) and laptop type (TypeName) are independent of each other.
- Alternative Hypothesis (H1): Laptop brand (Company) and laptop type (TypeName) are dependent (associated).

## Two Sample Z-Test

```python
# Two sample Z-test

from statsmodels.stats.weightstats import ztest

dell_price = Laptop_df1[Laptop_df1['Company'] == 'Dell']['Price']
hp_price   = Laptop_df1[Laptop_df1['Company'] == 'HP']['Price']

z_stat, p_val = ztest(dell_price, hp_price)

print("Z-statistic:", z_stat)
print("p-value:", p_val)

if p_val < 0.05:
    print("Reject H0 → Dell and HP prices are significantly different")
else:
    print("Fail to Reject H0 → No significant difference between Dell and HP")
```

**Test if the average laptop Price is different between two companies of Dell and HP.**
- H0: The mean price of Dell = mean price of HP.
- H1: The mean price of Dell != mean price of HP.
